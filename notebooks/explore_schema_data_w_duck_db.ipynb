## Loop though CSVs and read relation into dictionary

```python
from pathlib import Path
import duckdb

# 1. Define your Olist dataset directory
data_dir = Path("../data/raw")
oline_tables = {}

# 2. Loop through CSV data files
for csv_file in data_dir.glob("*.csv"):
    # Strip the file extension
    table_name_str = csv_file.stem
    table_name_str = table_name_str.replace("olist_", "") 
    table_name_str = table_name_str.replace("_dataset", "") 
    table_name_str = table_name_str + "_tmp"
    
    # Read the CSV file into a relation object and store it in the dictionary
    oline_tables[table_name_str] = duckdb.read_csv(csv_file)


oline_tables.items()
```

```python
# 3. Display table names
print("Olist Tables:")
for key in oline_tables.keys():
    print(f'- {key}')
```
