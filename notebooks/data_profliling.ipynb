## 1. Loop to generate data profiling reports for directory of Oline CSV files

```python
from pathlib import Path
import pandas as pd
from data_profiling import ProfileReport 

# Create list of CSV filenames
data_dir = Path("../data/raw")

# Iterate though CSV files and generate data profiles 
for filename in data_dir.glob("*.csv"):

    # Extract filename without extension
    filename_str = filename.stem
    
    if filename.suffix == '.csv':
        # Load CSV into pandas DataFrame
        df = pd.read_csv(filename)
        
        # Create a profiling report
        profile = ProfileReport(df, title=f'{filename_str} Profiling Report', explorative=True)
        
        # Export the profiling report to an HTML file
        profile.to_file(f'../data/profiles/{filename_str}_profile.html')
```

## 2. Combine each CSV dataset profile into a single multi-tab HTML report file

```python
from pathlib import Path
import html
import pandas as pd
from ydata_profiling import ProfileReport

# Define your Olist dataset directory and output directory
data_dir = Path("../data/raw")
out_dir = Path("../data/profiles")
out_dir.mkdir(parents=True, exist_ok=True)

sections = []

# Iterate through CSV files and generate data profiles
for csv_file in sorted(data_dir.glob("*.csv")):
    name = csv_file.stem.replace("olist_", "").replace("_dataset", "") + "_tbl"
    df = pd.read_csv(csv_file)
    report = ProfileReport(df, title=name.replace("_", " ").title(), minimal=True)
    sections.append((name, report.to_html()))
    print(f"Profiled {name}: {df.shape[0]:,} rows x {df.shape[1]} cols")

# Generate a single HTML page with tabs for each profile
tabs = "".join(
    f'<button onclick="show(\'{n}\')">{n}</button>' for n, _ in sections
)
frames = "".join(
    f'<iframe id="{n}" srcdoc="{html.escape(h)}" '
    f'style="display:none;width:100%;height:92vh;border:0"></iframe>'
    for n, h in sections
)
page = f"""<html><head><meta charset="utf-8"><title>Olist Profiles</title>
<style>button{{margin:4px;padding:6px 10px}}</style></head><body>
<div>{tabs}</div>{frames}
<script>
function show(id){{document.querySelectorAll('iframe').forEach(f=>f.style.display='none');
document.getElementById(id).style.display='block';}}
show('{sections[0][0]}');
</script></body></html>"""

# Write the combined HTML page to disk
(out_dir / "olist_profiles_combined.html").write_text(page, encoding="utf-8")
```
