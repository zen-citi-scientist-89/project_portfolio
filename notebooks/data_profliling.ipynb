## 1. Loop to generate data profiling reports for directory of Oline CSV files

```python
import os
import pandas as pd
from data_profiling import ProfileReport 

# Create list of CSV filenames
olist_filename_lst = os.listdir('../data/raw/')

# Iterate though CSV files and generate data profiles 
for filename in olist_filename_lst:
    if filename.endswith('.csv'):
        # Extract filename without extension
        filename_only = filename.split('.')[0]

        # Load CSV into pandas DataFrame
        df = pd.read_csv(f'../data/raw/{filename}')
        
        # Create a profiling report
        profile = ProfileReport(df, title=f'{filename_only} Profiling Report', explorative=True)
        
        # Export the profiling report to an HTML file
        profile.to_file(f'../data/profiles/{filename_only}_profile.html')
```

## 2. Combine each CSV dataset profile into a single HTML report file

```python
from pathlib import Path
import pandas as pd
from data_profiling import ProfileReport, compare

# 1. Define your Olist dataset directory
data_dir = Path("../data/raw")
reports = []

# 2. Process each CSV file into an in-memory profile structure
# Note: We use minimal=True to keep computation fast and light on RAM
for csv_file in data_dir.glob("*.csv"):
    df = pd.read_csv(csv_file)
    
    # Strip the file extension
    dataset_name = csv_file.stem 
    
    # Generate the profile object without writing it to disk
    report = ProfileReport(
        df, 
        title=dataset_name, 
        minimal=True, # Dramatically reduces memory usage for large pipelines
        explorative=False
    )
    reports.append(report)

# 3. Merge all reports side-by-side and export ONE single HTML file
if reports:
    print(f"Merging and profiling {len(reports)} Olist tables...")
    multi_report = compare(reports)
    multi_report.to_file("../data/profiles/olist_pipeline_comparison.html")
    print("Done! Open 'olist_pipeline_comparison.html' to evaluate everything.")
else:
    print("No CSV files found in the directory.")
```
