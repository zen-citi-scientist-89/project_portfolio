## Loop to generate data profiling reports for directory of Oline CSV files

```python
import os
import pandas as pd
from data_profiling import ProfileReport 

# Create list of CSV filenames
olist_filename_lst = os.listdir('../data/raw/')

# Iterate though CSV files and generate data profiles 
for filename in olist_filename_lst:
    if filename.endswith('.csv'):
        # Extract filename without extension
        filename_only = filename.split('.')[0]

        # Load CSV into pandas DataFrame
        df = pd.read_csv(f'../data/raw/{filename}')
        
        # Create a profiling report
        profile = ProfileReport(df, title=f'{filename_only} Profiling Report', explorative=True)
        
        # Export the profiling report to an HTML file
        profile.to_file(f'../data/profiles/{filename_only}_profile.html')
```
